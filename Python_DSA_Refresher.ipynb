```python
from prometheus_client import values

# prompts the user to enter the number of days
days = int(input("Enter the number of days: "))

#converts the days into hours, minutes then seconds
seconds = days * 24 * 60 * 60

print("Number of seconds:", seconds)
```

```python
import math
#prompts the user to enter the radius
radius = int(input("Enter the radius "))

#calculates the volume using the formula
volume = (4/3)*math.pi*(radius**3)

print("The volume is:", volume)
```

```python

#prompt the user to enter the length of a side
side = int(input("Enter the side length "))

#calculates the area and the perimeter of the square
perimeter = side*4
area = side*side

print("The area is:", area)
print("The perimeter is:", perimeter)
```

```python
def check_case(character):
    if character.isupper():
        print("The character is uppercase.")
    elif character.islower():
        print("The character is lowercase.")
    else:
        print("The input is not a letter.")


character = input("Enter a character: ")

check_case(character)
```

```python
x = 0
y = 20

while True:
    y = y - 4
    x = x + (2 / y)

    if y < 6:
        break

print(x)
```

```python
values = []

#input of 5 values
for i in range(5):
    value = float(input("Enter a value: "))
    values.append(value)

#calculate the average
average = sum(values) / len(values)

#display the values and average
print("values entered:", values)
print("The average is:", average)
```
